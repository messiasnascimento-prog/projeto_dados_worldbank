```python
import pandas as pd
import wbgapi as wb

# 1. Códigos dos indicadores
indicadores = {
    'SE.XPD.TOTL.GD.ZS': 'Gasto_Educacao_pct_PIB',
    'SL.UEM.TOTL.ZS': 'Taxa_Desemprego'
}

# 2. Região da América Latina (LCN) e anos atualizados de 2010 até 2025
regiao = 'LCN' 
anos = range(2000, 2026)

print("Iniciando a coleta de dados...")

# 3. Baixando da API do Banco Mundial
df = wb.data.DataFrame(list(indicadores.keys()), economy=regiao, time=anos, db=2)

# 4. Organizando a tabela
df = df.rename(index=indicadores)
df = df.T
df.index.name = 'Ano'
df = df.reset_index()

# Limpar o texto 'YR' da coluna de Ano
df['Ano'] = df['Ano'].str.replace('YR', '').astype(int)

# 5. Salvar em CSV (ponto e vírgula para o Excel do Brasil)
df.to_csv('dados_educacao_desemprego_AL.csv', index=False, sep=';', encoding='utf-8-sig')

print("Dados salvos com sucesso!")
df
```

```python
import matplotlib.pyplot as plt
import pandas as pd  

# Ler o arquivo CSV que criamos (garantindo o separador correto)
dados = pd.read_csv('dados_educacao_desemprego_AL.csv', sep=';')

# Configurar o tamanho do gráfico
plt.figure(figsize=(10, 5))

# --------------------------------------------------------------------------
# ONDE MUDOU: Criamos as tabelas limpas (sem NaNs) para a América Latina
# --------------------------------------------------------------------------
dados_educacao = dados.dropna(subset=['Gasto_Educacao_pct_PIB'])
dados_desemprego = dados.dropna(subset=['Taxa_Desemprego'])

# Plotar a linha de Gasto com Educação (repare que mudamos para dados_educacao)
plt.plot(dados_educacao['Ano'], dados_educacao['Gasto_Educacao_pct_PIB'], 
         label='Gasto em Educação (% do PIB)', marker='o', color='blue', linewidth=2)

# Plotar a linha de Taxa de Desemprego (repare que mudamos para dados_desemprego)
plt.plot(dados_desemprego['Ano'], dados_desemprego['Taxa_Desemprego'], 
         label='Taxa de Desemprego (%)', marker='s', color='red', linewidth=2)

# Adicionar títulos e legendas
plt.title('Evolução do Gasto com Educação vs Desemprego na América Latina (2000-2025)')
plt.xlabel('Ano')
plt.ylabel('Percentual (%)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# Mostrar o gráfico na tela
plt.show()
```

```python
# ==============================================================================
# 5. EXTRAÇÃO DE DADOS POR PAÍS (TOP 5 ECONOMIAS)
# ==============================================================================

# Definição dos 5 países da América Latina (códigos de 3 letras ISO)
paises = ['BRA', 'MEX', 'ARG', 'COL', 'CHL']

# Dicionário de tradução dos códigos de país para os nomes reais
nomes_paises = {
    'BRA': 'Brasil', 'MEX': 'México', 'ARG': 'Argentina', 
    'COL': 'Colômbia', 'CHL': 'Chile'
}

print("Acessando a API do Banco Mundial para os 5 países...")
# Buscando os mesmos indicadores, mas agora passando a lista de países em 'economy'
df_paises_bruto = wb.data.DataFrame(list(indicadores.keys()), economy=paises, time=anos, db=2)
print("Dados dos países extraídos com sucesso!")
```

```python
# ==============================================================================
# 6. PIPELINE DE TRATAMENTO E FILTRAGEM POR PAÍS
# ==============================================================================

df_paises = df_paises_bruto.copy()

# Resetar o índice duplo que a API cria para podermos manipular as colunas
df_paises = df_paises.reset_index()

# Renomear os códigos dos indicadores para nomes legíveis
df_paises['series'] = df_paises['series'].map(indicadores)
df_paises['economy'] = df_paises['economy'].map(nomes_paises)

# Organizar a tabela de forma que Anos e Países virem linhas estruturadas (Melt)
df_paises = df_paises.melt(id_vars=['economy', 'series'], var_name='Ano', value_name='Valor')

# Limpar o texto 'YR' da coluna de Ano e transformar em número inteiro
df_paises['Ano'] = df_paises['Ano'].str.replace('YR', '').astype(int)

# Pivotar a tabela para que as duas variáveis virem colunas separadas
df_paises = df_paises.pivot(index=['economy', 'Ano'], columns='series', values='Valor').reset_index()

# Salvar a nova base de dados detalhada por país em um arquivo CSV próprio
df_paises.to_csv('dados_top5_paises_AL.csv', index=False, sep=';', encoding='utf-8-sig')

print("Base de dados por países gerada com sucesso! Arquivo 'dados_top5_paises_AL.csv' salvo.")
df_paises.head(10) # Mostra as 10 primeiras linhas da nova estrutura
```

```python
# ==============================================================================
# 7. VISUALIZAÇÃO COMPARATIVA DOS 5 PAÍSES (SEM FALHAS NAS LINHAS)
# ==============================================================================

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Cores distintas para diferenciar cada país no gráfico
cores = {'Brasil': 'green', 'México': 'orange', 'Argentina': 'blue', 'Colômbia': 'purple', 'Chile': 'red'}

# Laço de repetição para plotar a linha de cada país
for pais, dados_pais in df_paises.groupby('economy'):
    
    # --------------------------------------------------------------------------
    # ONDE MUDA: Criamos os dados filtrados tirando os NaNs antes de plotar
    # --------------------------------------------------------------------------
    dados_educacao = dados_pais.dropna(subset=['Gasto_Educacao_pct_PIB'])
    dados_desemprego = dados_pais.dropna(subset=['Taxa_Desemprego'])
    
    # Gráfico 1: Gasto com Educação (Repare que mudamos para 'dados_educacao')
    ax1.plot(dados_educacao['Ano'], dados_educacao['Gasto_Educacao_pct_PIB'], 
             label=pais, marker='o', color=cores[pais], linewidth=2)
    
    # Gráfico 2: Taxa de Desemprego (Repare que mudamos para 'dados_desemprego')
    ax2.plot(dados_desemprego['Ano'], dados_desemprego['Taxa_Desemprego'], 
             label=pais, marker='s', color=cores[pais], linewidth=2)

# Configurações do Gráfico 1 (Educação)
ax1.set_title('Gasto em Educação (% do PIB) por País', fontsize=12, fontweight='bold')
ax1.set_xlabel('Ano')
ax1.set_ylabel('Percentual do PIB (%)')
ax1.grid(True, linestyle='--', alpha=0.5)
ax1.legend()

# Configurações do Gráfico 2 (Desemprego)
ax2.set_title('Taxa de Desemprego (%) por País', fontsize=12, fontweight='bold')
ax2.set_xlabel('Ano')
ax2.set_ylabel('Taxa de Desemprego (%)')
ax2.grid(True, linestyle='--', alpha=0.5)
ax2.legend()

plt.suptitle('Análise Comparativa das Principais Economias da América Latina (2000-2025)', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# 8. ABERTURA POR PAÍS: COMPARATIVO EM ESCALA UNIFICADA (%)
# ==============================================================================

import matplotlib.pyplot as plt

# Definição dos países analisados
lista_paises = ['Brasil', 'México', 'Argentina', 'Colômbia', 'Chile']

# Criar a estrutura com 5 gráficos verticais compartilhando o eixo X (Anos)
fig, axes = plt.subplots(5, 1, figsize=(14, 22), sharex=True)

for i, pais in enumerate(lista_paises):
    # Filtrar e ordenar a série histórica do país correspondente
    dados_pais = df_paises[df_paises['economy'] == pais].sort_values('Ano')
    ax = axes[i]
    
    # Tratamento para omitir valores nulos (NaN) e evitar quebras visuais na linha
    dados_educacao = dados_pais.dropna(subset=['Gasto_Educacao_pct_PIB'])
    dados_desemprego = dados_pais.dropna(subset=['Taxa_Desemprego'])
    
    # Plotar ambas as séries temporais diretamente no mesmo eixo (Escala Unificada)
    ax.plot(dados_educacao['Ano'], dados_educacao['Gasto_Educacao_pct_PIB'], 
            color='tab:blue', marker='o', linewidth=2, label='Gasto Educação (% PIB)')
    
    ax.plot(dados_desemprego['Ano'], dados_desemprego['Taxa_Desemprego'], 
            color='tab:red', marker='s', linestyle='--', linewidth=2, label='Taxa de Desemprego (%)')
    
    # Configurações estéticas do eixo Y do subgráfico
    ax.set_ylabel('Percentual (%)', fontsize=10)
    ax.grid(True, linestyle=':', alpha=0.6)
    
    # Identificação individual do país
    ax.set_title(f'{pais.upper()}: Dinâmica de Investimento em Educação e Mercado de Trabalho', 
                 fontsize=12, fontweight='bold', pad=10)
    
    # Renderizar a legenda no canto superior esquerdo de cada painel
    ax.legend(loc='upper left', fontsize=9)

# Configurações gerais do eixo do tempo (X) aplicado ao painel inferior
plt.xlabel('Ano', fontsize=12, fontweight='bold', labelpad=15)
plt.xticks(range(2000, 2027, 2))

# Título Principal do Relatório Visual
plt.suptitle('ABERTURA POR PAÍS: GASTO EM EDUCAÇÃO VS DESEMPREGO (2000-2025)\nAnálise Comparativa em Escala Unificada (%)', 
             fontsize=16, fontweight='bold', y=1.01)

# Ajuste de espaçamento para evitar sobreposição de textos
plt.tight_layout()
plt.show()
```
