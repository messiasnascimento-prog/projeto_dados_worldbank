```python
import matplotlib.pyplot as plt
import pandas as pd  # <-- Adicionamos essa linha para corrigir o erro!

# Ler o arquivo CSV que criamos (garantindo o separador correto)
dados = pd.read_csv('dados_educacao_desemprego_AL.csv', sep=';')

# Configurar o tamanho do gráfico
plt.figure(figsize=(10, 5))

# Plotar a linha de Gasto com Educação
plt.plot(dados['Ano'], dados['Gasto_Educacao_pct_PIB'], label='Gasto em Educação (% do PIB)', marker='o', color='blue')

# Plotar a linha de Taxa de Desemprego
plt.plot(dados['Ano'], dados['Taxa_Desemprego'], label='Taxa de Desemprego (%)', marker='s', color='red')

# Adicionar títulos e legendas
plt.title('Evolução do Gasto com Educação vs Desemprego na América Latina')
plt.xlabel('Ano')
plt.ylabel('Percentual (%)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# Mostrar o gráfico na tela
plt.show()
```
