```python
import pandas as pd
import wbgapi as wb

# 1. Códigos dos indicadores
indicadores = {
    'SE.XPD.TOTL.GD.ZS': 'Gasto_Educacao_pct_PIB',
    'SL.UEM.TOTL.ZS': 'Taxa_Desemprego'
}

# 2. Região da América Latina (LCN) e anos atualizados de 2010 até 2025
regiao = 'LCN' 
anos = range(2010, 2026)

print("Iniciando a coleta de dados...")

# 3. Baixando da API do Banco Mundial
df = wb.data.DataFrame(list(indicadores.keys()), economy=regiao, time=anos, db=2)

# 4. Organizando a tabela
df = df.rename(index=indicadores)
df = df.T
df.index.name = 'Ano'
df = df.reset_index()

# Limpar o texto 'YR' da coluna de Ano
df['Ano'] = df['Ano'].str.replace('YR', '').astype(int)

# 5. Salvar em CSV (ponto e vírgula para o Excel do Brasil)
df.to_csv('dados_educacao_desemprego_AL.csv', index=False, sep=';', encoding='utf-8-sig')

print("Dados salvos com sucesso!")
df
```
